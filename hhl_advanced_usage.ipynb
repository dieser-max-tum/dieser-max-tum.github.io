## Fast Tomography Methods

### 1. Direct Fidelity Estimation (DFE)
- **Speed**: Fastest
- **Measurements**: O(1/ε²)
- **Use case**: Quick validation

### 2. Classical Shadow Tomography
- **Speed**: Very fast
- **Measurements**: O(log d)
- **Use case**: Many observables

### 3. Randomized Benchmarking
- **Speed**: Fast
- **Measurements**: Minimal
- **Use case**: Gate characterization

### 4. Compressed Sensing
- **Speed**: Fast for low-rank
- **Measurements**: O(r log d)
- **Use case**: Large low-rank states

````python
import numpy as np
import matplotlib.pyplot as plt
from time import time

print('Fast tomography methods comparison')
print('='*60)

# Example: DFE is O(1/eps^2) faster than full tomography
n_qubits_range = range(2, 8)
full_tomo_measurements = [4**n for n in n_qubits_range]
dfe_measurements = [100] * len(n_qubits_range)  # Constant

plt.figure(figsize=(10, 6))
plt.semilogy(n_qubits_range, full_tomo_measurements, 'o-', label='Full Tomography', linewidth=2)
plt.semilogy(n_qubits_range, dfe_measurements, 's-', label='DFE', linewidth=2)
plt.xlabel('Number of Qubits', fontsize=12)
plt.ylabel('Measurements Required (log scale)', fontsize=12)
plt.title('Tomography Method Comparison', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('tomography_comparison.png', dpi=300)
plt.show()

print('\nRecommendation: Use DFE for fast benchmarking!')
````
